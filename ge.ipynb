```python
class GrammaticalEvolution:
    def __init__(self, grammar, xgboost_models, mandatory_ingredients, 
                 target_keywords, population_size=50, max_generations=100, 
                 mutation_rate=0.15, crossover_rate=0.8, max_wraps=3):
        """
        Initialize Grammatical Evolution for cocktail generation.
        
        Args:
            grammar: BNF grammar dictionary
            xgboost_models: dict with 'taste' and 'keywords' XGBoost models
            mandatory_ingredients: list of ingredients user has (can be empty)
            target_keywords: list of desired keywords (e.g., ['fruity', 'sweet'])
            population_size: number of individuals in population
            max_generations: max number of evolution iterations
            mutation_rate: probability of mutation
            crossover_rate: probability of crossover
            max_wraps: maximum number of times to wrap around genome
        """
        self.grammar = grammar
        self.xgboost_taste = xgboost_models['taste']
        self.xgboost_keywords = xgboost_models['keywords']
        self.mandatory_ingredients = set(mandatory_ingredients) if mandatory_ingredients else set()
        self.target_keywords = set(target_keywords)
        self.population_size = population_size
        self.max_generations = max_generations
        self.mutation_rate = mutation_rate
        self.crossover_rate = crossover_rate
        self.max_wraps = max_wraps
        self.codon_size = 256
        
        # Track best solutions
        self.best_individual = None
        self.best_fitness = float('-inf')
        self.fitness_history = []
        self.best_recipe = None
        
    def initialize_population(self):
        """Create initial random population."""
        population = []
        for _ in range(self.population_size):
            # Random genome of integers (codons)
            genome_length = random.randint(40, 120)
            genome = [random.randint(0, self.codon_size - 1) for _ in range(genome_length)]
            population.append(genome)
        return population
    
    def map_genome_to_recipe(self, genome):
        """Map genome to cocktail recipe using grammar."""
        try:
            recipe_parts = []
            current_symbol = '<recipe>'
            genome_index = 0
            wraps = 0
            expansion_count = 0
            max_expansions = 200  # Prevent infinite loops
            
            while expansion_count < max_expansions:
                # Check if we're done (no more non-terminals)
                if '<' not in current_symbol:
                    break
                
                # Find first non-terminal
                start = current_symbol.find('<')
                if start == -1:
                    break
                
                end = current_symbol.find('>', start)
                if end == -1:
                    break
                
                non_terminal = current_symbol[start:end+1]
                
                if non_terminal not in self.grammar:
                    # Unknown non-terminal, skip it
                    current_symbol = current_symbol[:start] + current_symbol[end+1:]
                    continue
                
                # Get production rules for this non-terminal
                choices = self.grammar[non_terminal]
                if not choices:
                    current_symbol = current_symbol[:start] + current_symbol[end+1:]
                    continue
                
                # Use codon to select production
                codon = genome[genome_index % len(genome)]
                choice_index = codon % len(choices)
                production = choices[choice_index]
                
                # Replace non-terminal with production
                current_symbol = current_symbol[:start] + production + current_symbol[end+1:]
                
                genome_index += 1
                expansion_count += 1
                
                if genome_index >= len(genome):
                    wraps += 1
                    
                if wraps >= self.max_wraps:
                    break
            
            # Parse the final string to extract recipe
            recipe = self._parse_recipe_string(current_symbol)
            
            # Add mandatory ingredients if not present
            for ingredient in self.mandatory_ingredients:
                if ingredient not in recipe:
                    recipe[ingredient] = 30.0  # Default percentage
            
            # Normalize to 100%
            recipe = self._normalize_recipe(recipe)
            
            return recipe
            
        except Exception as e:
            # Return empty recipe on error
            return {}
    
    def _parse_recipe_string(self, recipe_string):
        """Parse recipe string into ingredient:percentage dict."""
        recipe = {}
        parts = recipe_string.split('|')
        
        for part in parts:
            part = part.strip()
            if ':' in part:
                try:
                    ingredient, percentage = part.split(':', 1)
                    ingredient = ingredient.strip()
                    percentage = float(percentage.strip())
                    
                    if ingredient in ALL_INGREDIENTS and percentage > 0:
                        recipe[ingredient] = percentage
                except:
                    continue
        
        return recipe
    
    def _normalize_recipe(self, recipe):
        """Normalize recipe so percentages sum to 100%."""
        if not recipe:
            return {}
        
        total = sum(recipe.values())
        if total == 0:
            return {}
        
        normalized = {ingredient: (pct / total) * 100 
                     for ingredient, pct in recipe.items()}
        
        return normalized
    
    def recipe_to_features(self, recipe):
        """Convert recipe dict to feature vector for XGBoost."""
        if not recipe:
            return None
        
        # Create feature dictionary with all ingredients as percentages
        features = {}
        
        for ingredient in ALL_INGREDIENTS:
            col_name = ingredient + '_pct'
            features[col_name] = recipe.get(ingredient, 0.0)
        
        return features
    
    def set_expected_features(self, feature_columns):
        """
        Set the expected feature columns from your trained model.
        Call this after loading your XGBoost models.
        
        Args:
            feature_columns: list of column names your model expects
        """
        self.expected_features = feature_columns
    
    def calculate_fitness(self, recipe):
        """Calculate fitness score for a recipe."""
        if not recipe or len(recipe) == 0:
            return -1000
        
        # Check mandatory ingredients are present
        for ingredient in self.mandatory_ingredients:
            if ingredient not in recipe:
                return -1000
        
        # Convert to features
        features = self.recipe_to_features(recipe)
        if features is None:
            return -1000
        
        # Create DataFrame for XGBoost
        feature_df = pd.DataFrame([features])

        if hasattr(self, 'expected_features'):
            # Add missing columns with 0
            for col in self.expected_features:
                if col not in feature_df.columns:
                    feature_df[col] = 0.0
            # Reorder to match exactly
            feature_df = feature_df[self.expected_features]
        
        try:
            # Get predictions from both models
            taste_rating = self.xgboost_taste.predict(feature_df)[0]
            keyword_predictions = self.xgboost_keywords.predict(feature_df)[0]
            
            # keyword_predictions is array of 10 values (one per keyword)
            keyword_dict = {kw: keyword_predictions[i] 
                          for i, kw in enumerate(KEYWORD_COLUMNS)}
            
            fitness = 0
            
            # ================================================================
            # KEYWORD MATCHING (Most Important)
            # ================================================================
            for keyword in self.target_keywords:
                if keyword in keyword_dict:
                    # High reward for matching target keywords
                    fitness += keyword_dict[keyword] * 100
            
            # Penalty for strong unwanted keywords
            for keyword in KEYWORD_COLUMNS:
                if keyword not in self.target_keywords:
                    fitness -= keyword_dict[keyword] * 10
            
            # ================================================================
            # PRACTICAL CONSTRAINTS
            # ================================================================
            
            # Number of ingredients (prefer 3-7 ingredients)
            num_ingredients = len(recipe)
            if num_ingredients < 2:
                fitness -= 100  # Too simple
            elif num_ingredients > 10:
                fitness -= (num_ingredients - 10) * 10  # Too complex
            elif 3 <= num_ingredients <= 7:
                fitness += 20  # Ideal complexity bonus
            
            # Check for base spirit (at least one major spirit)
            base_spirits = ['gin', 'vodka', 'rum', 'whiskey', 'bourbon_whiskey', 
                          'rye_whiskey_50%_abv', 'scotch_whisky', 'cognac_(brandy)',
                          'blanco_tequila', 'reposado_tequila', 'mezcal', 
                          'light_gold_rum_1-3yo', 'light_white_rum', 'aged_rum_(6-10yr)']
            
            has_base = any(spirit in recipe for spirit in base_spirits)
            if not has_base:
                fitness -= 50  # Penalty for no base spirit
            
            # Base spirit should be 30-60% of total
            base_spirit_total = sum(recipe.get(spirit, 0) for spirit in base_spirits)
            if base_spirit_total < 25:
                fitness -= 30  # Too weak
            elif base_spirit_total > 70:
                fitness -= 20  # Too strong
            elif 35 <= base_spirit_total <= 55:
                fitness += 15  # Ideal range bonus
            
            # Discourage tiny amounts (< 3%)
            for pct in recipe.values():
                if pct < 3:
                    fitness -= 5
            
            # ================================================================
            # BALANCE AND DIVERSITY
            # ================================================================
            
            # Reward diversity in ingredient types
            has_juice = any('juice' in ing for ing in recipe.keys())
            has_syrup = any('syrup' in ing for ing in recipe.keys())
            has_bitter = any('bitter' in ing for ing in recipe.keys())
            
            diversity_score = sum([has_juice, has_syrup, has_bitter])
            fitness += diversity_score * 5
            
            return fitness
            
        except Exception as e:
            print(f"Error in fitness calculation: {e}")
            return -1000
    
    def tournament_selection(self, population, fitnesses, tournament_size=3):
        """Select individual using tournament selection."""
        tournament_indices = random.sample(range(len(population)), tournament_size)
        tournament_fitnesses = [fitnesses[i] for i in tournament_indices]
        winner_index = tournament_indices[tournament_fitnesses.index(max(tournament_fitnesses))]
        return population[winner_index].copy()
    
    def crossover(self, parent1, parent2):
        """Single-point crossover."""
        if random.random() > self.crossover_rate:
            return parent1.copy(), parent2.copy()
        
        point = random.randint(1, min(len(parent1), len(parent2)) - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2
    
    def mutate(self, genome):
        """Mutate genome by randomly changing codons."""
        mutated = genome.copy()
        for i in range(len(mutated)):
            if random.random() < self.mutation_rate:
                mutated[i] = random.randint(0, self.codon_size - 1)
        return mutated
    
    def evolve(self, verbose=True):
        """Run the evolutionary algorithm."""
        population = self.initialize_population()
        
        for generation in range(self.max_generations):
            # Evaluate fitness
            recipes = [self.map_genome_to_recipe(genome) for genome in population]
            fitnesses = [self.calculate_fitness(recipe) for recipe in recipes]
            
            # Track best
            max_fitness = max(fitnesses)
            max_index = fitnesses.index(max_fitness)
            
            if max_fitness > self.best_fitness:
                self.best_fitness = max_fitness
                self.best_individual = population[max_index].copy()
                self.best_recipe = recipes[max_index].copy()
            
            self.fitness_history.append(max_fitness)
            
            if verbose:
                print(f"Gen {generation + 1}/{self.max_generations} | "
                      f"Best Fitness: {max_fitness:.2f} | "
                      f"Best Overall: {self.best_fitness:.2f} | "
                      f"Ingredients: {len(recipes[max_index])}")
            
            # Create next generation
            new_population = []
            
            # Elitism - keep top 10% of population
            elite_count = max(1, self.population_size // 10)
            elite_indices = sorted(range(len(fitnesses)), 
                                 key=lambda i: fitnesses[i], 
                                 reverse=True)[:elite_count]
            for idx in elite_indices:
                new_population.append(population[idx].copy())
            
            # Fill rest with offspring
            while len(new_population) < self.population_size:
                # Selection
                parent1 = self.tournament_selection(population, fitnesses)
                parent2 = self.tournament_selection(population, fitnesses)
                
                # Crossover
                child1, child2 = self.crossover(parent1, parent2)
                
                # Mutation
                child1 = self.mutate(child1)
                child2 = self.mutate(child2)
                
                new_population.extend([child1, child2])
            
            population = new_population[:self.population_size]
        
        return self.best_recipe, self.best_fitness

```

```python
def create_cocktail_grammar():
    """Create BNF grammar for cocktail recipes."""
    
    # Define percentage ranges for different ingredient types
    spirit_percentages = ['25', '30', '35', '40', '45', '50', '55', '60']
    liqueur_percentages = ['10', '15', '20', '25', '30']
    juice_percentages = ['10', '15', '20', '25', '30', '35']
    syrup_percentages = ['5', '10', '15', '20']
    bitter_percentages = ['1', '2', '3', '4', '5']
    modifier_percentages = ['10', '20', '30', '40']
    
    # Major ingredient categories
    spirits = ['gin', 'vodka', 'light_gold_rum_1-3yo', 'light_white_rum', 
               'bourbon_whiskey', 'rye_whiskey_50%_abv', 'scotch_whisky', 
               'cognac_(brandy)', 'blanco_tequila', 'reposado_tequila', 
               'mezcal', 'aged_rum_(6-10yr)']
    
    liqueurs = ['triple_sec', 'orange_curaçao', 'amaretto_liqueur', 
                'coffee_liqueur', 'elderflower_liqueur', 'maraschino_liqueur',
                'chartreuse_green', 'chartreuse_yellow', 'grand_marnier',
                'apricot_liqueur', 'limoncello_liqueur']
    
    juices = ['lemon_juice', 'lime_juice', 'orange_juice', 'pineapple_juice',
              'pink_grapefruit_juice', 'cranberry_juice', 'apple_juice']
    
    syrups = ['sugar_syrup_(2:1)', 'agave_syrup', 'honey_and_water_syrup',
              'grenadine_syrup', 'orgeat_syrup', 'brown_sugar_syrup']
    
    bitters = ['aromatic_bitters', 'orange_bitters', 'angostura_bitters']
    
    vermouths = ['dry_vermouth', 'rosso/sweet_vermouth', 'bianco/blanco_vermouth']
    
    modifiers = ['soda_(club_soda)_water', 'tonic_water', 'ginger_ale', 
                 'ginger_beer', 'prosecco']
    
    grammar = {
        '<recipe>': [
            '<base>',
            '<base> | <juice>',
            '<base> | <sweetener>',
            '<base> | <juice> | <sweetener>',
            '<base> | <liqueur> | <juice>',
            '<base> | <liqueur> | <sweetener>',
            '<base> | <juice> | <sweetener> | <bitter>',
            '<base> | <liqueur> | <juice> | <sweetener>',
            '<base> | <secondary> | <juice> | <sweetener>',
            '<base> | <vermouth> | <bitter>',
            '<base> | <liqueur> | <juice> | <sweetener> | <bitter>',
            '<base> | <juice> | <sweetener> | <modifier>',
        ],
        
        '<base>': [f'{spirit}:{pct}' 
                   for spirit in spirits 
                   for pct in spirit_percentages],
        
        '<secondary>': [f'{spirit}:{pct}' 
                       for spirit in spirits 
                       for pct in ['10', '15', '20', '25']],
        
        '<liqueur>': [f'{liq}:{pct}' 
                     for liq in liqueurs 
                     for pct in liqueur_percentages],
        
        '<juice>': [f'{juice}:{pct}' 
                   for juice in juices 
                   for pct in juice_percentages],
        
        '<sweetener>': [f'{syrup}:{pct}' 
                       for syrup in syrups 
                       for pct in syrup_percentages],
        
        '<bitter>': [f'{bitter}:{pct}' 
                    for bitter in bitters 
                    for pct in bitter_percentages],
        
        '<vermouth>': [f'{vermouth}:{pct}' 
                      for vermouth in vermouths 
                      for pct in ['15', '20', '25', '30']],
        
        '<modifier>': [f'{mod}:{pct}' 
                      for mod in modifiers 
                      for pct in modifier_percentages],
    }
    
    return grammar
```

```python
import joblib

xgboost_models = {
    'taste': joblib.load('taste_rating_model.pkl'),
    'keywords': joblib.load('keywords_model.pkl')
}

X_train = pd.read_csv('cocktail_dataset_final.csv')
# Remove target columns if they're in the dataset
feature_columns = [col for col in X_train.columns 
                  if col not in ['bittersweet', 'citrus', 'creamy', 'floral', 
                                'fruity', 'herbal', 'nutty', 'savoury', 
                                'spicy', 'sweet', 'strength_rating', 
                                'taste_rating', 'rating', 'recipe_id']]

# 2. Define user inputs
mandatory_ingredients = ['rum', 'lime_juice']  # or [] for no requirements
target_keywords = ['fruity', 'sweet']

# 3. Create grammar
grammar = create_cocktail_grammar()

# 4. Initialize and run GE
ge = GrammaticalEvolution(
    grammar=grammar,
    xgboost_models=xgboost_models,
    mandatory_ingredients=mandatory_ingredients,
    target_keywords=target_keywords,
    population_size=50,
    max_generations=100,
    mutation_rate=0.15,
    crossover_rate=0.8
)

ge.set_expected_features(feature_columns)

# 5. Evolve the best recipe
best_recipe, best_fitness = ge.evolve()

# 6. Display results
print("\\n" + "="*60)
print("BEST COCKTAIL RECIPE:")
print("="*60)
for ingredient, percentage in sorted(best_recipe.items(), 
                                     key=lambda x: x[1], 
                                     reverse=True):
    print(f"{ingredient}: {percentage:.1f}%")
print(f"\\nFitness Score: {best_fitness:.2f}")

```
