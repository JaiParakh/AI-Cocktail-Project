```python
class EnhancedDiffordsScraper:
    """Extract data from JSON-LD AND strength/taste ratings from images"""

    def __init__(self):
        self.session = requests.Session()
        self.session.headers.update({
          'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36',
          'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
          'Accept-Language': 'en-US,en;q=0.9',
          'Accept-Encoding': 'gzip, deflate, br',
          'Upgrade-Insecure-Requests': '1',
          'Sec-Fetch-Dest': 'document',
          'Sec-Fetch-Mode': 'navigate',
          'Sec-Fetch-Site': 'none',
          'Cookie': r'diffords_webapp=2ofangpatjcs193lais0ct77r4; diffords_subscribe_upsell_counter=1; pg_last_pageview_data={"cohortId":"4459","cohortUpdatedAt":"2512171245","configId":19506}; pg_buildfile=251217-ec6-nc-b0038d4b76294bd9b3171fb2aae5adca; pg_unq_cohort_key=4459:2512171245; _ga=GA1.1.1524274016.1767191083; pg_geo={"country":"IE","region":"LK","ip":"193.1.98.143"}; pg_ip=193.1.98.143; pg_vignettePercent=1; pg_anchorPercent=1; pg_sideRailPercent=1; pg_auto_ads=0; pg_chaser=1; pg_canonical_session=www.diffordsguide.com/cocktails; pg_lazy=1; _sharedID=198ca399-724f-49e3-820d-edc88c7e74ea; _sharedID_cst=zix7LPQsHA%3D%3D; CloudFront-Policy=eyJTdGF0ZW1lbnQiOlt7IlJlc291cmNlIjoiaHR0cHM6Ly9hc3NldHMuZGlmZm9yZHNndWlkZS5jb20vKiIsIkNvbmRpdGlvbiI6eyJEYXRlTGVzc1RoYW4iOnsiQVdTOkVwb2NoVGltZSI6MTc2NzI3NzQ4NH19fV19; CloudFront-Signature=etZt2ePXDDjv9zMcxAaeaq8MAWd2vay6gvzOUqwPw1JjlICE9xqUMu7mme6Cg3Dr44YslrhJ32n-6cRTbN-et~qZVROHzUCd1pJYYdnaYhJUvhK37gqeKqKLu2XKJeyGKXhcxSD630mzK9vNYZ4uAIkXhrEw5gzC51Ajez7EfKQDTPqRsHH36SlYM66aF0YY0jyy6Byq5VSSfaZn9Ae46ORZZDhJBF~nkz6rkgNKmkvdac4-XJ8axF8kPqCQkM8BvJ~ZJQG3kL6n2HlkP~Srt4IrWmuuadVNCtfPiMFVOGnRKWWMsvd3jEIu0QgewVhwqMZe1hrXU2aPb5eEBkvdlQ__; CloudFront-Key-Pair-Id=K18QK78KYYZSIQ; pg_analytics=0; pg_mm2_cookieA=e4e31a71-7fc6-41d7-8596-f96aeed64a39; pg_sessionId=f51b5d20-1f74-4a0b-b034-8f703f8484aa; pg_session_depth=1; FCCDCF=%5Bnull%2Cnull%2Cnull%2C%5B%22CQdTAsAQdTAsAEsACBENCKFsAP_gAEPgAAwIK1IB_C7EbCFCiDJ3IKMEMAhHABBAYsAwAAYBAwAADBIQIAQCgkEYBASAFCACCAAAKASBAAAgCAAAAUAAIAAFAABAAAwAIBAIIAAAgAAAAEAIAAAACIAAEQCAAAAEAEAAkAgAAAIASAAAAAAAAACBAAAAAAAAAAAAAAAABAAAAQAAQAAAAAAAiAAAAAAAABAIAAAAAAAAAAAAAAAAAAAAAAgAAAAAAAAAABAAAAAAAQR2QD-F2I2EKFEGCuQUYIYBCuACAAxYBgAAwCBgAAGCQgQAgFJIIkCAEAIEAAEAAAQAgCAABQEBAAAIAAAAAqAACAABgAQCAQQIABAAAAgIAAAAAAEQAAIgEAAAAIAIABABAAAAQAkAAAAAAAAAECAAAAAAAAAAAAAAAAAAAAAEABgAAAAAABEAAAAAAAACAQIAAA%22%2C%222~61.89.122.161.184.196.230.314.442.445.494.550.576.827.1029.1033.1046.1047.1051.1097.1126.1166.1301.1342.1415.1725.1765.1942.1958.1987.2068.2072.2074.2107.2213.2219.2223.2224.2328.2331.2387.2416.2501.2567.2568.2575.2657.2686.2778.2869.2878.2908.2920.2963.3005.3023.3100.3126.3219.3234.3235.3253.3309.3731.6931.8931.13731.15731.33931~dv.%22%2C%226E32C4E1-33A6-4DF7-B4EE-0EC9C9F57EF7%22%5D%2Cnull%2Cnull%2C%5B%5B32%2C%22%5B%5C%22c6d2d133-2926-40a1-9a70-434c581189b8%5C%22%2C%5B1767191083%2C906000000%5D%5D%22%5D%5D%5D; pg_anchorUnitFromConfig=0; pg_beacon=1; pg_mm2_cookie_a=e4e31a71-7fc6-41d7-8596-f96aeed64a39; pg_session_id=f51b5d20-1f74-4a0b-b034-8f703f8484aa; pg_tc=sample; pg_tc_dd=true; pg_quick_check=true; pg_ua=Mozilla/5.0 (Windows NT 10.0 Win64 x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36; pg_inner_width=1536; pg_tc_device=desktop; pg_latency_before_tc=194; pg_tc_platform=Windows; pg_tc_platform_version=19.0.0; pg_tc_model=; pg_pl=15; pg_lava=; pg_bot_percent=24.8; pg_bot_reason=lnb; pg_bot_model=1; pg_tc_response_time=302; __gads=ID=a39e5a470f08ff44:T=1767191088:RT=1767191088:S=ALNI_MY_s60TZw_FVJGVY-zuxhZ0URikjg; __gpi=UID=000012db14295e8a:T=1767191088:RT=1767191088:S=ALNI_MazUnuWo5FpD3aKz5KQkCs9YCQgWA; __eoi=ID=5e8cb6f024ee70fc:T=1767191088:RT=1767191088:S=AA-AfjakI9QDbpnZII3AldYIHniz; _ga_G28QGN42XB=GS2.1.s1767191083$o1$g0$t1767191089$j60$l0$h0; pg_pv_time_1=1044'
        })

    def get_jsonld_data(self, url: str) -> Dict:
        """Get JSON-LD data from a cocktail URL"""
        try:
            print(f"📥 Fetching: {url}")

            # Get page
            response = self.session.get(url, timeout=10)
            response.raise_for_status()
            soup = BeautifulSoup(response.content, 'html.parser')

            # 1. Find JSON-LD script tag
            script_tag = soup.find('script', type='application/ld+json')

            if not script_tag:
                print(f"⚠️ No JSON-LD found for {url}")
                return {'error': 'No JSON-LD data', 'url': url}

            # Parse JSON
            data = json.loads(script_tag.string)

            # Extract recipe data (might be nested in @graph)
            recipe_data = data
            if '@graph' in data:
                for item in data['@graph']:
                    if item.get('@type') == 'Recipe':
                        recipe_data = item
                        break

            # 2. Extract strength and taste ratings from images
            strength_data = self._extract_strength_taste_ratings(soup)

            # 3. Combine all data
            combined_data = {
                **recipe_data,
                **strength_data,  # Add strength/taste ratings
                'scraped_url': url,
                'cocktail_id': self._extract_id(url),
                'scrape_timestamp': pd.Timestamp.now(),
                'scrape_success': True
            }

            print(f"✅ Found: {recipe_data.get('name', 'Unknown')}")
            print(f"   Strength: {strength_data.get('strength_rating_raw', 'Not found')}")
            print(f"   Taste: {strength_data.get('taste_rating_raw', 'Not found')}")

            return combined_data

        except Exception as e:
            print(f"❌ Error: {str(e)}")
            return {
                'error': str(e),
                'url': url,
                'scrape_success': False,
                'scrape_timestamp': pd.Timestamp.now()
            }

    def _extract_id(self, url: str) -> str:
        """Extract cocktail ID from URL"""
        match = re.search(r'/recipe/(\d+)/', url)
        return match.group(1) if match else 'unknown'

    def _extract_strength_taste_ratings(self, soup: BeautifulSoup) -> Dict:
        """
        Extract strength and taste ratings from image alt attributes
        Looks for patterns like:
        - <img src="/svg/range/9.svg" alt="Strength 9/10">
        - <img src="/svg/range/7.svg" alt="Sweet to sour 7/10">
        """
        strength_data = {
            'strength_rating_raw': '',
            'taste_rating_raw': '',
            'strength_found': False,
            'taste_found': False
        }

        # Find all images with range/svg in src
        range_images = soup.find_all('img', src=re.compile(r'/svg/range/'))

        for img in range_images:
            alt_text = img.get('alt', '')

            if not alt_text:
                continue

            # Check for strength rating
            if 'strength' in alt_text.lower():
                strength_data['strength_rating_raw'] = alt_text
                strength_data['strength_found'] = True

            # Check for taste rating (sweet/sour/dry)
            elif any(keyword in alt_text.lower() for keyword in ['sweet', 'sour', 'dry']):
                strength_data['taste_rating_raw'] = alt_text
                strength_data['taste_found'] = True

        # Alternative search: Look for divs with specific classes
        if not strength_data['strength_found'] or not strength_data['taste_found']:
            # Look for divs with 'cell py-2' class (from your HTML example)
            cell_divs = soup.find_all('div', class_='cell py-2')

            for div in cell_divs:
                img = div.find('img')
                if img and img.get('alt'):
                    alt_text = img['alt']

                    if 'strength' in alt_text.lower() and not strength_data['strength_found']:
                        strength_data['strength_rating_raw'] = alt_text
                        strength_data['strength_found'] = True

                    elif any(keyword in alt_text.lower() for keyword in ['sweet', 'sour', 'dry']) and not strength_data['taste_found']:
                        strength_data['taste_rating_raw'] = alt_text
                        strength_data['taste_found'] = True

        return strength_data
```

```python
def flatten_for_table(jsonld_data: Dict) -> Dict:
    """Flatten JSON-LD data into a single table row with strength/taste ratings"""

    if 'error' in jsonld_data:
        return {
            'url': jsonld_data.get('url', ''),
            'error': jsonld_data.get('error', 'Unknown error'),
            'scrape_success': False,
            'scrape_timestamp': jsonld_data.get('scrape_timestamp', '')
        }

    # Basic info
    flat = {
        'cocktail_id': jsonld_data.get('cocktail_id', ''),
        'scraped_url': jsonld_data.get('scraped_url', ''),
        'name': jsonld_data.get('name', ''),
        'headline': jsonld_data.get('headline', ''),
        'description': jsonld_data.get('description', ''),
        'recipe_category': jsonld_data.get('recipeCategory', ''),
        'recipe_cuisine': jsonld_data.get('recipeCuisine', ''),
        'date_published': jsonld_data.get('datePublished', ''),
        'scrape_success': jsonld_data.get('scrape_success', True),
        'scrape_timestamp': jsonld_data.get('scrape_timestamp', ''),

        # New: Strength and taste ratings (raw strings)
        'strength_rating_raw': jsonld_data.get('strength_rating_raw', ''),
        'taste_rating_raw': jsonld_data.get('taste_rating_raw', ''),
        'strength_found': jsonld_data.get('strength_found', False),
        'taste_found': jsonld_data.get('taste_found', False)
    }

    # Keywords
    keywords = jsonld_data.get('keywords', [])
    flat['keywords'] = ', '.join(keywords) if isinstance(keywords, list) else str(keywords)

    # Recipe yield
    recipe_yield = jsonld_data.get('recipeYield', '')
    if isinstance(recipe_yield, list):
        flat['recipe_yield'] = ', '.join(recipe_yield)
    else:
        flat['recipe_yield'] = recipe_yield

    # Timing
    flat['total_time'] = jsonld_data.get('totalTime', '')

    # Nutrition
    nutrition = jsonld_data.get('nutrition', {})
    if isinstance(nutrition, dict):
        flat['calories'] = nutrition.get('calories', '')
    else:
        flat['calories'] = ''

    # Rating
    rating = jsonld_data.get('aggregateRating', {})
    if isinstance(rating, dict):
        flat['rating_value'] = rating.get('ratingValue', '')
        flat['rating_count'] = rating.get('ratingCount', '')

    # Instructions count and summary
    instructions = jsonld_data.get('recipeInstructions', [])
    if isinstance(instructions, list):
        flat['instructions_count'] = len(instructions)
        # Join first 3 instructions
        instruction_texts = []
        for instr in instructions[:3]:
            if isinstance(instr, dict):
                instruction_texts.append(instr.get('text', ''))
            elif isinstance(instr, str):
                instruction_texts.append(instr)
        flat['instructions_summary'] = ' | '.join(instruction_texts)[:200]

    # Ingredients - add as separate columns
    ingredients = jsonld_data.get('recipeIngredient', [])
    flat['ingredients_count'] = len(ingredients)

    # Add each ingredient in separate columns (up to 15)
    for i, ing in enumerate(ingredients[:15]):
        flat[f'ingredient_{i+1}'] = ing

    # Calculate total ml (very basic)
    total_ml = 0
    for ing in ingredients:
        if isinstance(ing, str):
            # Try to extract ml value
            match = re.search(r'([\d\.]+)\s*ml', ing, re.IGNORECASE)
            if match:
                try:
                    total_ml += float(match.group(1))
                except:
                    pass
    flat['estimated_total_ml'] = total_ml

    # Add all remaining JSON-LD fields (except those already processed)
    processed_keys = set(flat.keys())
    for key, value in jsonld_data.items():
        if key not in processed_keys and not key.startswith('@'):
            if isinstance(value, (str, int, float, bool)):
                flat[key] = value
            elif isinstance(value, (list, dict)):
                flat[key] = str(value)

    return flat
```

```python
# %% [markdown]
# # Difford's JSON-LD Data Extractor (Updated)
# Now with strength & taste rating extraction from image alt text

# %%


# %%


# %% [markdown]
# ## Enhanced JSON-LD Parser with Strength/Taste Extraction

# %%


# %% [markdown]
# ## Data Flattener (Updated with Strength/Taste)

# %%


# %% [markdown]
# ## Test with Negroni Example

# %%
# Test URL


# Show what we got
print("\n=== EXTRACTED DATA ===")
print(f"Name: {jsonld_data.get('name')}")
print(f"Strength rating: {jsonld_data.get('strength_rating_raw', 'Not found')}")
print(f"Taste rating: {jsonld_data.get('taste_rating_raw', 'Not found')}")
print(f"Ingredients: {len(jsonld_data.get('recipeIngredient', []))}")

# Show ingredients
print("\n🧪 Ingredients:")
for i, ing in enumerate(jsonld_data.get('recipeIngredient', [])):
    print(f"   {i+1}. {ing}")

# Show rating
rating = jsonld_data.get('aggregateRating', {})
if rating:
    print(f"\n⭐ Rating: {rating.get('ratingValue')}/5 ({rating.get('ratingCount')} ratings)")

# %% [markdown]
# ## Flatten and Create DataFrame

# %%
# Flatten the data
flat_data = flatten_for_table(jsonld_data)

# Create DataFrame
df = pd.DataFrame([flat_data])

print("\n=== FLATTENED DATA TABLE ===")
print(f"Table shape: {df.shape}")
print(f"Columns: {len(df.columns)}")

# Show important columns including new strength/taste
important_cols = [
    'cocktail_id', 'name', 'ingredients_count', 'estimated_total_ml',
    'strength_rating_raw', 'taste_rating_raw',
    'rating_value', 'rating_count', 'calories'
]

available_cols = [col for col in important_cols if col in df.columns]

print("\n📊 Summary Table:")
print(df[available_cols].to_string(index=False))

# Show all columns
print(f"\n📋 All {len(df.columns)} columns:")
cols_per_line = 3
columns = df.columns.tolist()
for i in range(0, len(columns), cols_per_line):
    line_columns = columns[i:i+cols_per_line]
    print("  " + "  ".join([f"{col:30}" for col in line_columns]))

# %% [markdown]
# ## Batch Scraping Function

# %%
def scrape_cocktail_list(urls: List[str], delay: float = 2.0) -> pd.DataFrame:
    """Scrape multiple cocktail URLs with strength/taste extraction"""
    scraper = EnhancedDiffordsScraper()
    all_data = []

    print(f"Starting batch scrape of {len(urls)} URLs...")

    for i, url in enumerate(urls):
        print(f"\n[{i+1}/{len(urls)}] Processing...")

        # Get data
        jsonld_data = scraper.get_jsonld_data(url)

        # Flatten
        flat_data = flatten_for_table(jsonld_data)
        all_data.append(flat_data)

        # Delay to be nice to the server
        if i < len(urls) - 1:
            time.sleep(delay)

    # Create DataFrame
    df = pd.DataFrame(all_data)

    # Stats
    success = df['scrape_success'].sum() if 'scrape_success' in df.columns else 0
    strength_found = df['strength_found'].sum() if 'strength_found' in df.columns else 0
    taste_found = df['taste_found'].sum() if 'taste_found' in df.columns else 0

    print(f"\n✅ Batch complete!")
    print(f"   Success rate: {success}/{len(urls)} ({success/len(urls)*100:.1f}%)")
    print(f"   Strength ratings found: {strength_found}/{len(urls)}")
    print(f"   Taste ratings found: {taste_found}/{len(urls)}")

    return df

# %% [markdown]
# ## URL Generator

# %%
def generate_urls(start_id: int = 1, count: int = 20) -> List[str]:
    """Generate cocktail URLs based on ID numbers"""
    base_url = "https://www.diffordsguide.com/cocktails/recipe/{}/"
    urls = []

    for i in range(start_id, start_id + count):
        urls.append(base_url.format(i))

    return urls

# Generate some test URLs
print("Generating test URLs...")
test_urls = generate_urls(1390, 3)  # IDs 1390-1392
print(f"Generated {len(test_urls)} URLs:")
for url in test_urls:
    print(f"  • {url}")

# %% [markdown]
# ## Run Batch Test

# %%
# Run small batch test
batch_df = scrape_cocktail_list(test_urls, delay=3.0)
print(f"\nBatch data shape: {batch_df.shape}")

# Show batch results
print("\n=== BATCH RESULTS ===")
print(batch_df[['cocktail_id', 'name', 'strength_rating_raw', 'taste_rating_raw']].to_string(index=False))

# %% [markdown]
# ## Save Data to CSV

# %%
def save_to_csv(df: pd.DataFrame, filename: str = "cocktails_with_ratings.csv"):
    """Save data to CSV file"""
    df.to_csv(filename, index=False, encoding='utf-8')
    print(f"✅ Saved to {filename}")
    print(f"   Rows: {len(df)}")
    print(f"   Columns: {len(df.columns)}")

    # Show column count by type
    strength_cols = [col for col in df.columns if 'strength' in col.lower()]
    taste_cols = [col for col in df.columns if 'taste' in col.lower()]
    ingredient_cols = [col for col in df.columns if 'ingredient' in col.lower()]

    print(f"   Strength columns: {len(strength_cols)}")
    print(f"   Taste columns: {len(taste_cols)}")
    print(f"   Ingredient columns: {len(ingredient_cols)}")

    # Try to download in Colab
    try:
        from google.colab import files
        files.download(filename)
        print("   Download started...")
    except:
        print("   File saved locally (not in Colab)")

# Save the batch data
save_to_csv(batch_df, "test_batch_with_ratings.csv")

# %% [markdown]
# ## View Specific Strength/Taste Data

# %%
def view_rating_data(df: pd.DataFrame):
    """Show strength and taste rating distribution"""
    print("=== STRENGTH & TASTE RATING ANALYSIS ===")

    # Strength ratings
    strength_ratings = df['strength_rating_raw'].value_counts().head(10)
    print("\n📊 Top 10 Strength Ratings:")
    for rating, count in strength_ratings.items():
        print(f"   {rating}: {count}")

    # Taste ratings
    taste_ratings = df['taste_rating_raw'].value_counts().head(10)
    print("\n👅 Top 10 Taste Ratings:")
    for rating, count in taste_ratings.items():
        print(f"   {rating}: {count}")

    # Missing data
    missing_strength = df['strength_rating_raw'].isna().sum() + (df['strength_rating_raw'] == '').sum()
    missing_taste = df['taste_rating_raw'].isna().sum() + (df['taste_rating_raw'] == '').sum()

    print(f"\n⚠️ Missing Data:")
    print(f"   Strength ratings missing: {missing_strength}/{len(df)}")
    print(f"   Taste ratings missing: {missing_taste}/{len(df)}")

# View rating data from batch
if not batch_df.empty:
    view_rating_data(batch_df)

# %% [markdown]
# ## Complete Usage Example

# %%
print("""
=== COMPLETE USAGE ===

1. SINGLE COCKTAIL:
   scraper = EnhancedDiffordsScraper()
   data = scraper.get_jsonld_data("URL")
   flat = flatten_for_table(data)
   df = pd.DataFrame([flat])

2. BATCH SCRAPE:
   urls = generate_urls(1000, 100)  # IDs 1000-1099
   batch_df = scrape_cocktail_list(urls, delay=2.0)
   save_to_csv(batch_df, "100_cocktails.csv")

3. NEW DATA YOU GET:
   • strength_rating_raw: "Strength 9/10"
   • taste_rating_raw: "Sweet to sour 7/10"
   • strength_found: True/False
   • taste_found: True/False

4. TIPS:
   • Start with small batches to test
   • Use delay=2-3 seconds
   • Check 'strength_found' and 'taste_found' columns
   • Raw strings preserved - no processing!
""")

# %% [markdown]
# ## Quick Test Function

# %%
def quick_test(url: str):
    """Quick test function"""
    scraper = EnhancedDiffordsScraper()
    data = scraper.get_jsonld_data(url)

    if 'error' in data:
        print(f"Error: {data['error']}")
        return None

    # Show key info
    print(f"\n🎉 Successfully scraped: {data.get('name')}")
    print(f"   ID: {data.get('cocktail_id')}")
    print(f"   Strength: {data.get('strength_rating_raw', 'Not found')}")
    print(f"   Taste: {data.get('taste_rating_raw', 'Not found')}")

    # Create DataFrame
    flat = flatten_for_table(data)
    df = pd.DataFrame([flat])

    return df

# Test with another URL
print("\n" + "="*50)
print("Quick test with another cocktail...")
another_url = "https://www.diffordsguide.com/cocktails/recipe/1/absinthe-drip"
df_test = quick_test(another_url)

if df_test is not None:
    print(f"\n✅ DataFrame ready with {len(df_test.columns)} columns")
    print("Columns with 'rating' in name:")
    rating_cols = [col for col in df_test.columns if 'rating' in col.lower()]
    for col in rating_cols:
        print(f"   • {col}: {df_test[col].iloc[0]}")
```
